```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder.appName("PySparkPractice").master("local[*]") .getOrCreate()

sc = spark.sparkContext
```

```python

numbers = [5, 12, 7, 20, 3, 16, 9, 24, 10, 1]

numbers_rdd = sc.parallelize(numbers)

filtered_rdd = numbers_rdd.filter(lambda x: x > 10)

result_rdd = filtered_rdd.map(lambda x: x * 2)

print(result_rdd.collect())
```

```python
student_marks = [
    ("Arun", 78),
    ("Divya", 92),
    ("Kavin", 45),
    ("Meena", 66),
    ("Ravi", 35),
    ("Priya", 88)
]

students_rdd = sc.parallelize(student_marks)

filtered_students = students_rdd.filter(lambda student: student[1] >= 50)

result_rdd = filtered_students.map(lambda student: student[0] + " scored " + str(student[1]))

print(result_rdd.collect())
```

```python
employee_data = [
    (1, "Arun", "Male", 24, "Chennai", 28000),
    (2, "Divya", "Female", 27, "Bengaluru", 42000),
    (3, "Kavin", "Male", 22, "Chennai", 25000),
    (4, "Meena", "Female", 30, "Hyderabad", 52000),
    (5, "Ravi", "Male", 28, "Bengaluru", 38000),
    (6, "Priya", "Female", 25, "Chennai", 46000)
]

columns = ["id", "name", "gender", "age", "city", "salary"]

employee_df = spark.createDataFrame(employee_data, columns)

employee_df.show()
```

```python
from pyspark.sql.functions import col

employee_df.filter(col("salary") > 40000).show()
```

```python
product_data = [
    (101, "Notebook", 40, 5),
    (102, "Pen", 10, 20),
    (103, "Pencil", 5, 30),
    (104, "School Bag", 650, 3),
    (105, "Eraser", 8, 15)
]

columns = ["product_id", "product_name", "price", "quantity"]

product_df = spark.createDataFrame(product_data, columns)

product_df.show()
```

```python
from pyspark.sql.functions import col

result_df = product_df.withColumn(
    "total_value",
    col("price") * col("quantity")
)

result_df.show()
```

```python
movie_data = [
    ("Leo", "Action", 250, 7.8),
    ("Hi Nanna", "Drama", 75, 8.2),
    ("Jailer", "Action", 200, 7.1),
    ("Good Night", "Comedy", 30, 8.0),
    ("Dragon", "Comedy", 100, 7.5),
    ("Amaran", "Drama", 150, 8.3)
]

columns = ["movie_name", "category", "collection_crore", "rating"]

movie_df = spark.createDataFrame(movie_data, columns)

movie_df.select("movie_name", "category", "rating").show()
```

```python
from pyspark.sql.functions import col

filtered_movies = movie_df.filter(col("rating") >= 8.0)

result_df = filtered_movies.withColumn(
    "collection_with_10_percent_bonus",
    col("collection_crore") * 1.10
)

result_df.show()
```
